```python
import gwsus as sus
import example_01_SR2_data as data
import symengine as se
```

## Declare suspension instance

```python
PAY = sus.Suspension("PAY")
```

## Declare ground and masses

```python
GND = sus.Ground( 'GND', data.GND )

MIR = sus.RigidBody( 'MIR', data.MIR )
RM  = sus.RigidBody( 'RM', data.RM )
IM  = sus.RigidBody( 'IM', data.IM )
IRM = sus.RigidBody( 'IRM', data.IRM )

BF  = sus.GASFilter( 'BF', data.BF )
F1  = sus.GASFilter( 'F1', data.F1 )
```

## Declare wires

```python
wire_MIR = sus.Wire( data.wire_MIR )
wire_RM = sus.Wire( data.wire_RM )
wire_IM = sus.Wire( data.wire_IM )
wire_IRM = sus.Wire( data.wire_IRM )
wire_BF = sus.Wire( data.wire_BF )
wire_F1 = sus.Wire( data.wire_F1 )
```

```python
type( wire_F1.length )
```

## Connect ground and masses with wires 

```python
PAY.connect( GND , 'for_F1' , F1 , 'for_GND' , wire_F1 )
PAY.connect( F1, 'for_BF' , BF , 'for_F1' , wire_BF )
PAY.connect( BF, 'for_IM', IM, 'for_BF', wire_IM )
PAY.connect( BF, 'for_IRM', IRM , 'for_BF' , wire_IRM )
PAY.connect( IM , 'for_MIR' , MIR , 'for_IM' , wire_MIR )
PAY.connect( IM , 'for_RM' , RM , 'for_IM' , wire_RM )

GND.show()
PAY.update()
```
