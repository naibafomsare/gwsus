```python
import ast
```

```python
code = """
import symengine as se
a = se.Rational(1/3)
b = se.sympify(2/7)
c = se.Rational(1, 3)
"""

tree = ast.parse(code)
for node in ast.walk(tree):
    if isinstance(node, ast.Call):
        # Check if calling Rational or sympify with a '/' division argument
        for arg in node.args:
            if isinstance(arg, ast.BinOp) and isinstance(arg.op, ast.Div):
                print(f"Warning on line {node.lineno}: Division '/' evaluated before passing to {ast.dump(node.func)}")
```
