```python
import gwsus as sus
import SR2_data as data
import symengine as se
```

## Declare suspension instance

```python
PAY = sus.Suspension("PAY")
```

## Declare ground and masses

```python
GND = sus.Ground( PAY , 'GND' , data = data.GND )
MIR  = sus.RigidBody( PAY , 'MIR' , data = data.MIR )
RM   = sus.RigidBody( PAY , 'RM' , data = data.RM )
IM   = sus.RigidBody( PAY , 'IM' , data = data.IM )
IRM   = sus.RigidBody( PAY , 'IRM' , data = data.IRM )
BF   = sus.RigidBody( PAY , 'BF' , data = data.BF )
F1   = sus.RigidBody( PAY , 'F1' , data = data.F1 )
```

## Declare wires

```python
wire_MIR = sus.Wire( PAY , data = data.wire_MIR )
wire_RM = sus.Wire( PAY , data = data.wire_RM )
wire_IM = sus.Wire( PAY , data = data.wire_IM )
wire_IRM = sus.Wire( PAY , data = data.wire_IRM )
wire_BF = sus.Wire( PAY , data = data.wire_BF )
wire_F1 = sus.Wire( PAY , data = data.wire_F1 )
```

## Connect ground and masses with wires 

```python
# PAY.connect( GND , 'for_IM' , IM , 'for_GND' , wire_IM )
# PAY.connect( GND , 'for_IRM' , IRM , 'for_GND' , wire_IRM )
# PAY.connect( GND , 'for_BF' , BF , 'for_GND' , wire_BF )

PAY.connect( GND , 'for_F1' , F1 , 'for_GND' , wire_F1 )
PAY.connect( F1, 'for_BF' , BF , 'for_F1' , wire_BF )

PAY.connect( BF, 'for_IM', IM, 'for_BF', wire_IM )
PAY.connect( BF, 'for_IRM', IRM , 'for_BF' , wire_IRM )
PAY.connect( IM , 'for_MIR' , MIR , 'for_IM' , wire_MIR )
PAY.connect( IM , 'for_RM' , RM , 'for_IM' , wire_RM )

GND.show()
PAY.update()
```
